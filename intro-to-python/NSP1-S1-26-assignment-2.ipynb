#### 1. Write a Python program using Matplotlib to plot a simple line graph showing the relationship  between two lists of data. The first list contains the numbers [1, 2, 3, 4, 5] and the second list  contains the corresponding squares [1, 4, 9, 16, 25].

```python
import matplotlib.pyplot as plt

# Data lists
x_values = [1, 2, 3, 4, 5]
y_values = [1, 4, 9, 16, 25]

# Create the plot
plt.plot(x_values, y_values, marker='o', color='b', linestyle='-')

# Add labels and a title
plt.title('Numbers vs. Their Squares')
plt.xlabel('Numbers')
plt.ylabel('Squares')

# Display grid lines for readability
plt.grid(True)

# Render the graph
plt.show()
```

####4. Write a Python program to create a 1x2 grid of subplots. The first subplot should show a simple  line plot with data [1, 2, 3, 4] and [10, 20, 30, 40]. The second subplot should display a bar plot  with the same x-axis values and corresponding y-axis values [5, 10, 15, 20].

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y_line = [10, 20, 30, 40]
y_bar = [5, 10, 15, 20]

# Create a 1x2 grid of subplots (1 row, 2 columns)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# First subplot: Line Plot
ax1.plot(x, y_line, marker='o', color='blue', linestyle='-')
ax1.set_title('Line Plot')
ax1.set_xlabel('X-axis')
ax1.set_ylabel('Y-axis')
ax1.grid(True)

# Second subplot: Bar Plot
ax2.bar(x, y_bar, color='green')
ax2.set_title('Bar Plot')
ax2.set_xlabel('X-axis')
ax2.set_ylabel('Y-axis')
ax2.grid(True, axis='y')

# Adjust spacing so titles and labels do not overlap
plt.tight_layout()

# Display the plots
plt.show()
```

#### 6. Create a Python program that defines a custom function to calculate the factorial of a number.  The function should take one integer as an argument and return its factorial. Sample Input -> 5 Sample Output -> The factorial of 5 is: 120

```python
def calculate_factorial(n):
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    return factorial

# Input
num = 5

# Output
print(f"The factorial of {num} is: {calculate_factorial(num)}")
```

####9. You are working as a data analyst in a college where the academic coordinator wants to evaluate how students perform across different subjects and sections. You are given a dataset containing Student, Subject, Marks, and Section. Using Seaborn, you need to create a catplot() to visualize the average marks for each subject, while using hue="Section" to compare performance between different sections. You should experiment with both "bar" and "point" plot types to understand how each represents the data differently. Enhance the visualization by adding a meaningful title such as “Subject-wise Performance by Section”, choosing an appropriate color palette, and increasing the plot size for clarity. After generating the plots, analyze them to determine which subject shows the highest variation in marks across sections and interpret what this suggests about student performance.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Sample Dataset Creation
data = {
    'Student': [f'Student_{i}' for i in range(1, 25)],
    'Subject': ['Math', 'Science', 'English', 'History'] * 6,
    'Section': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'] * 3,
    'Marks': [88, 92, 75, 70, 65, 90, 78, 72,
              85, 95, 80, 68, 60, 88, 76, 74,
              90, 91, 77, 71, 58, 89, 79, 73]
}

df = pd.DataFrame(data)

# Set Seaborn visual style
sns.set_theme(style="whitegrid")

# 2. Catplot - Type "bar"
bar_plot = sns.catplot(
    data=df,
    x="Subject",
    y="Marks",
    hue="Section",
    kind="bar",
    palette="Set2",
    height=5,
    aspect=1.5,
    errorbar=None  # Set to 'sd' or None depending on preference
)
bar_plot.fig.suptitle("Subject-wise Performance by Section (Bar Plot)", y=1.03)
bar_plot.set_axis_labels("Subject", "Average Marks")

# 3. Catplot - Type "point"
point_plot = sns.catplot(
    data=df,
    x="Subject",
    y="Marks",
    hue="Section",
    kind="point",
    palette="Set2",
    height=5,
    aspect=1.5,
    markers=["o", "s"],
    linestyles=["-", "--"]
)
point_plot.fig.suptitle("Subject-wise Performance by Section (Point Plot)", y=1.03)
point_plot.set_axis_labels("Subject", "Average Marks")
plt.show()
```

####10. You are analyzing sales data for a retail company operating across multiple regions. The dataset includes Product, Region, and Sales. Your task is to use catplot(kind="bar") in Seaborn to compare product sales across regions, using hue="Product" to differentiate between products. Additionally, apply col="Region" to create separate visualizations for each region, enabling clearer comparisons. Customize the plot using the "Set2" color palette and adjust the height and aspect to improve readability. Once the visualization is complete, analyze the patterns to identify which product performs consistently well across all regions and explain the possible reasons behind this trend.

```python
import random
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Generate Synthetic Data with Random Product Names
products = ['Laptop', 'X-Box', 'Headphones']
regions = ['North', 'South', 'East', 'West']

data = []
# Set seed for reproducible sales ranges while keeping product selection clear
random.seed(42)

for region in regions:
    for product in products:
        # Assign different sales ranges to demonstrate performance trends
        if product == 'Laptop':
            sales = random.randint(115, 135)  # High performance across regions
        elif product == 'X-Box':
            sales = random.randint(80, 95)    # Moderate performance
        else:
            sales = random.randint(50, 70)    # Lower performance

        data.append({'Region': region, 'Product': product, 'Sales': sales})

df = pd.DataFrame(data)

# Set Seaborn theme
sns.set_theme(style="whitegrid")

# 2. Plot using catplot
g = sns.catplot(
    data=df,
    x="Product",
    y="Sales",
    hue="Product",
    col="Region",
    kind="bar",
    palette="Set2",
    height=4,
    aspect=0.8,
    legend=False
)

# Customization
g.fig.subplots_adjust(top=0.8)
g.fig.suptitle("Regional Sales Comparison by Product", fontsize=16, fontweight='bold')
g.set_axis_labels("Product", "Sales ($ in Thousands)")
g.set_titles(col_template="Region: {col_name}")

plt.show()
```
