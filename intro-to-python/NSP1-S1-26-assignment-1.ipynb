### 1. Given a number, find the sum of its digits.Take the number as an input from the user

```python
while True:
    user_input = input("Enter an integer: ")

    # Strip the leading negative sign if it exists for validation purposes
    clean_input = user_input[1:] if user_input.startswith('-') else user_input

    # Combined Check: Rejects empty inputs OR inputs containing non-digits
    if user_input.strip() == "" or not clean_input.isdigit():
        print("Invalid input! Please enter a valid whole number (no letters, spaces, or decimals).")
    else:
        break

# Initialize the sum
digit_sum = 0

digit_sum = sum(int(char) for char in clean_input)

print(f"The sum of the digits is: {digit_sum}")
```

### 2. Given a number, check whether the given number is an Armstrong number or not. A positive integer is called an Armstrong number of order n if:
abcd... = an + bn + cn + dn + ...
Example: 153 = 1*1*1 + 5*5*5 + 3*3*3
153 is an Armstrong number of order 3

```python
def get_valid_integer(prompt_message):
    """Repeatedly prompts the user until a valid positive integer is entered."""
    while True:
        user_input = input(prompt_message)
        # Rejects empty inputs OR inputs containing non-digits
        if user_input.strip() == "" or not user_input.isdigit():
            print("Invalid input! Please enter a valid, positive whole number.")
        else:
            return user_input

# 1. Get the number and order using the helper function
target_number = get_valid_integer("Enter a positive integer to check: ")
order_input = get_valid_integer("Enter the order (n): ")
n = int(order_input)

# 2. Compute the Armstrong sum in one line
armstrong_sum = sum(int(char) ** n for char in target_number)

# 3. Final Comparison
if armstrong_sum == int(target_number):
    print(f"Yes! {target_number} is an Armstrong number of order {n}.")
else:
    print(f"No. {target_number} is NOT an Armstrong number of order {n}.")
```

### 3. Given a string, write a python function to check if it is  palindrome or not. A string is said to be palindrome if the reverse of the string is the same as string. For example, “malayalam” is a palindrome, but “music” is not a palindrome.

```python
# Lambda function that cleans the text and checks if it equals its reverse
is_palindrome = lambda text: "".join(text.split()).lower() == "".join(text.split()).lower()[::-1]

user_input = input("Enter a string to check for palindrome: ")

if is_palindrome(user_input):
    print(f'Yes! "{user_input}" is a palindrome.')
else:
    print(f'No. "{user_input}" is NOT a palindrome.')
```

### 4. Given an array which may contain duplicates, print all elements and their frequencies

```python
def print_frequencies(arr):
    # Initialize an empty dictionary to store counts
    frequency_map = {}

    # Iterate through the array and update counts
    for element in arr:
        # dict.get(element, 0) returns the current count, or 0 if it's the first time seeing it
        frequency_map[element] = frequency_map.get(element, 0) + 1

    # Print the elements and their frequencies
    print("\nElement -> Frequency")
    print("--------------------")
    for element, count in frequency_map.items():
        print(f"{element} -> {count}")

# --- Testing the Function ---
# A sample array containing duplicates
sample_array = [1, 2, 8, 3, 2, 2, 2, 5, 1, 8, 3, 3]

print_frequencies(sample_array)
```

### 6. Given a number n, write a function to print all prime factors of n. For example, if the input number is 12, then output should be “2 2 3”.

```python
def prime_factors(n):
    """Generates all prime factors of n without external libraries."""
    if n <= 1:
        return

    # 1. Pull out all factors of 2
    while n % 2 == 0:
        yield 2
        n //= 2

    # 2. Check odd factors up to the square root of n
    factor = 3
    while factor <= n ** 0.5:
        while n % factor == 0:
            yield factor
            n //= factor
        factor += 2

    # 3. If n is still greater than 2, the remaining n is prime
    if n > 2:
        yield n

# --- Testing the Pythonic Approach ---
number = 12

# Using a list comprehension or join to print the yielded values cleanly
result = " ".join(str(f) for f in prime_factors(number))
print(f"Prime factors of {number}: {result}")
```

### 7. Given two numbers n and r, find the value of nCr (binomial coefficient: nCr = (n!) / (r! * (n-r)!))

```python
def find_ncr(n, r):
    # Edge Cases: nCr is 0 if r is out of bounds
    if r < 0 or r > n:
        return 0

    # Choosing the smaller r cuts our calculation loop steps in half!
    if r > n - r:
        r = n - r

    # Calculate using a sleek generator/reduction concept
    result = 1
    for i in range(1, r + 1):
        # We multiply by the top factor and divide by the bottom factor in one step
        result = result * (n - r + i) // i

    return result

# --- Testing the Function ---
n_val = 5
r_val = 2

# 5C2 = (5 * 4) / (2 * 1) = 10
print(f"{n_val}C{r_val} = {find_ncr(n_val, r_val)}")
```

### 7. Searching: Given a sorted array arr[] of n elements, write a function to search a given element x in arr[]. Do it using linear and binary search techniques.

```python
def linear_search(arr, x):
    """Scans the array sequentially to find element x."""
    for index, element in enumerate(arr):
        if element == x:
            return index  # Return the index where element was found

    return -1  # Return -1 if element is not in the array

def binary_search(arr, x):
    """Uses two pointers to divide and conquer the sorted array."""
    left = 0
    right = len(arr) - 1

    while left <= right:
        # Calculate the middle index (integer division)
        mid = (left + right) // 2

        # Check if x is present at mid
        if arr[mid] == x:
            return mid

        # If x is greater, ignore the left half
        elif arr[mid] < x:
            left = mid + 1

        # If x is smaller, ignore the right half
        else:
            right = mid - 1

    return -1  # Return -1 if element is not present

# A sorted sample array
sorted_array = [2, 5, 8, 12, 16, 23, 38, 45, 56, 72, 91]
target = 23

print(f"Target to find: {target}\n")

# Linear Search Test
linear_result = linear_search(sorted_array, target)
print(f"[Linear Search] Element found at index: {linear_result}")

# Binary Search Test
binary_result = binary_search(sorted_array, target)
print(f"[Binary Search] Element found at index: {binary_result}")
```

###8. Input a text file (containing 1 or more paragraphs of English text) from the user, parse this file to display the frequency of occurrence of each word in this text file. Find the 3 most frequent words as well

```python
import string

def analyze_word_frequency(file_path):
    word_counts = {}

    try:
        # 1. Open and read the file safely
        with open(file_path, 'r', encoding='utf-8') as file:
            for line in file:
                # 2. Clean punctuation and normalize case line-by-line
                # string.punctuation contains: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
                cleaned_line = line.translate(str.maketrans('', '', string.punctuation)).lower()

                # Split the line into individual words based on whitespace
                words = cleaned_line.split()

                # 3. Populate the frequency dictionary
                for word in words:
                    word_counts[word] = word_counts.get(word, 0) + 1

    except FileNotFoundError:
        print(f"Error: The file at '{file_path}' could not be found.")
        return

    if not word_counts:
        print("The file is empty or contains no valid words.")
        return

    # 4. Display the frequency of each unique word
    print("\n--- Word Frequencies ---")
    for word, count in sorted(word_counts.items()):
        print(f"'{word}': {count}")

    # 5. Find the 3 most frequent words
    # We sort the dictionary items by their value (count) in descending order
    top_3 = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)[:3]

    print("\n--- Top 3 Most Frequent Words ---")
    for index, (word, count) in enumerate(top_3, 1):
        print(f"{index}. '{word}' appears {count} times")

analyze_word_frequency("/content/exercise_8_sample_file.txt")
```

###9. A hospital is planning to digitize its patient record system to improve efficiency in managing patient information and billing. As a Python developer, you are tasked with building a simple system to process and analyze patient data. You are given a collection of patient details, where each record contains the patient’s name, age, disease, and total bill amount.
*  Store and organize patient data using appropriate data structures.
*  Calculate the number of patients affected by each disease.
*   Identify the patient with the highest bill amount
*   Compute the average bill and display patients whose bill exceeds the average
*   Ensure that duplicate patient entries are removed using suitable techniques







```python
def process_hospital_data(raw_data):
    # --- Step 1: Remove Duplicate Entries ---
    # We track unique patients using a tuple of (name, age, disease) to catch exact clones
    seen_patients = []
    cleaned_patients = []

    for record in raw_data:
        # Create a unique identifier key for the patient
        patient_key = (record["name"], record["age"], record["disease"])

        if patient_key not in seen_patients:
            seen_patients.append(patient_key)
            cleaned_patients.append(record)

    if not cleaned_patients:
        print("No patient records to process.")
        return

    print(f"Processing complete. Removed {len(raw_data) - len(cleaned_patients)} duplicate record(s).\n")

    # --- Step 2: Calculate Disease Frequencies & Highest Bill & Totals ---
    disease_counts = {}
    highest_bill_patient = cleaned_patients[0]
    total_bill_sum = 0

    for patient in cleaned_patients:
        # Count diseases
        disease = patient["disease"]
        disease_counts[disease] = disease_counts.get(disease, 0) + 1

        # Track highest bill
        if patient["bill"] > highest_bill_patient["bill"]:
            highest_bill_patient = patient

        # Accumulate total for average
        total_bill_sum += patient["bill"]

    # --- Step 3: Compute Average Bill ---
    total_patients = len(cleaned_patients)
    average_bill = total_bill_sum / total_patients

    # --- Step 4: Display Analytics and Results ---

    # Print Disease Analytics
    print("--- Disease Distribution ---")
    for disease, count in disease_counts.items():
        print(f"• {disease}: {count} patient(s)")
    print("-" * 30)

    # Print Highest Bill Earner
    print("--- Highest Billing Patient ---")
    print(f"Name: {highest_bill_patient['name']}")
    print(f"Disease: {highest_bill_patient['disease']}")
    print(f"Bill Amount: ${highest_bill_patient['bill']:.2f}")
    print("-" * 30)

    # Print Average & Above Average Bills
    print(f"--- Patients with Above-Average Bills (Avg: ${average_bill:.2f}) ---")
    for patient in cleaned_patients:
        if patient["bill"] > average_bill:
            print(f"• {patient['name']} (Bill: ${patient['bill']:.2f}, Disease: {patient['disease']})")


# --- Testing Data Setup ---
raw_patient_records = [
    {"name": "Alice Smith", "age": 34, "disease": "Flu", "bill": 250.0},
    {"name": "Bob Jones", "age": 45, "disease": "Diabetes", "bill": 1200.0},
    {"name": "Alice Smith", "age": 34, "disease": "Flu", "bill": 250.0},       # Duplicate entry 1
    {"name": "Charlie Brown", "age": 60, "disease": "Heart Disease", "bill": 5000.0},
    {"name": "Diana Prince", "age": 29, "disease": "Flu", "bill": 150.0},
    {"name": "Evan Wright", "age": 52, "disease": "Diabetes", "bill": 3100.0},
    {"name": "Bob Jones", "age": 45, "disease": "Diabetes", "bill": 1200.0}     # Duplicate entry 2
]

# Run the program
process_hospital_data(raw_patient_records)
```

```python
# --- 1. Structured Tabular Inline CSV Dataset (28 Records + 1 Duplicate) ---
# This mimics an actual CSV file structure, complete with a header and 29 lines of data.
inline_csv_data = """order_id,restaurant,customer_id,item_name,quantity,unit_price
201,Wok & Roll,CUST25,Lo Mein,4,11.00
201,Wok & Roll,CUST25,Spring Rolls,4,2.50
202,Taco Fiesta,CUST15,Burrito,2,9.00
203,Wok & Roll,CUST10,Kung Pao Chicken,1,14.50
203,Wok & Roll,CUST10,Kung Pao Chicken,1,14.50
204,Wok & Roll,CUST13,Kung Pao Chicken,1,14.50
204,Wok & Roll,CUST13,Spring Rolls,1,2.50
204,Wok & Roll,CUST13,Lo Mein,4,11.00
205,Taco Fiesta,CUST12,Burrito,4,9.00
205,Taco Fiesta,CUST12,Churros,3,4.50
206,Burger Joint,CUST12,Fries,4,3.00
206,Burger Joint,CUST12,Cheeseburger,3,8.50
207,Sushi Zen,CUST20,California Roll,2,7.00
207,Sushi Zen,CUST20,Miso Soup,1,3.00
208,Pizza Palace,CUST14,Pepperoni Pizza,2,18.00
208,Pizza Palace,CUST14,Garlic Knots,4,4.00
208,Pizza Palace,CUST14,Soda,3,2.50
209,Taco Fiesta,CUST22,Beef Taco,4,3.50
210,Burger Joint,CUST19,Milkshake,2,5.00
211,Sushi Zen,CUST11,Salmon Nigiri,3,5.50
211,Sushi Zen,CUST11,Tempura Udon,1,13.00
212,Wok & Roll,CUST17,Fried Rice,2,12.00
213,Pizza Palace,CUST24,Margherita Pizza,1,15.50
214,Taco Fiesta,CUST16,Quesadilla,2,8.00
215,Burger Joint,CUST10,Cheeseburger,4,8.50
215,Burger Joint,CUST10,Onion Rings,2,4.00
216,Sushi Zen,CUST18,California Roll,3,7.00
217,Pizza Palace,CUST15,Pepperoni Pizza,1,18.00
218,Wok & Roll,CUST21,Spring Rolls,3,2.50"""


def analyze_inline_food_orders(csv_string):
    restaurant_revenue = {}
    item_frequencies = {}

    # Track structural duplicates using a composite unique key: (order_id, item_name)
    seen_entries = []
    duplicate_row_count = 0
    total_processed_rows = 0

    # Split the raw multiline string by newlines to simulate reading file lines
    lines = csv_string.strip().split("\n")

    # Isolate the header row (index 0)
    header = lines[0]

    # Process the data lines (from index 1 onward)
    for line in lines[1:]:
        cleaned_line = line.strip()
        if not cleaned_line:
            continue

        total_processed_rows += 1

        # Manual CSV cell parsing using basic string splits
        row = cleaned_line.split(',')

        # Schema Mapping:
        # row[0]=order_id, row[1]=restaurant, row[3]=item_name, row[4]=quantity, row[5]=unit_price
        order_id = row[0]
        restaurant = row[1]
        item_name = row[3]
        quantity = int(row[4])
        unit_price = float(row[5])

        # Check for structural duplicates
        entry_key = (order_id, item_name)
        if entry_key in seen_entries:
            duplicate_row_count += 1
            continue  # Bypass processing this duplicated row

        # Unique record row configuration
        seen_entries.append(entry_key)

        # Calculate financial metrics for this line item
        line_total = quantity * unit_price

        # Accumulate the total revenue metrics for the restaurant
        restaurant_revenue[restaurant] = restaurant_revenue.get(restaurant, 0.0) + line_total

        # Accumulate quantitative demand frequencies for the items
        item_frequencies[item_name] = item_frequencies.get(item_name, 0) + quantity

    # Check if data exists
    if not restaurant_revenue:
        print("Data processing stopped: The dataset contains no records.")
        return

    # --- Analytics & Metric Extraction Engine ---

    # Find the top-performing restaurant based on total revenue generated
    top_restaurant = None
    max_revenue = -1.0
    for rest, rev in restaurant_revenue.items():
        if rev > max_revenue:
            max_revenue = rev
            top_restaurant = rest

    # Identify the most frequently ordered item based on cumulative quantities
    most_frequent_item = None
    max_freq = -1
    for item, freq in item_frequencies.items():
        if freq > max_freq:
            max_freq = freq
            most_frequent_item = item

    # --- Display Structured Insights Report ---
    print("==========================================")
    print("      DATA INGESTION & PIPELINE REPORT    ")
    print("==========================================")
    print(f"Total Rows Scanned   : {total_processed_rows}")
    print(f"Duplicate Rows Found : {duplicate_row_count}")
    print(f"Valid Rows Processed : {total_processed_rows - duplicate_row_count}")
    print("-" * 42)

    print("\n--- Total Revenue Generated Per Restaurant ---")
    for rest, rev in restaurant_revenue.items():
        print(f"• {rest.ljust(15)}: ${rev:,.2f}")
    print("-" * 42)

    print("\n--- Platform Business Winner ---")
    print(f"Top Restaurant : {top_restaurant}")
    print(f"Total Revenue  : ${max_revenue:,.2f}")
    print("-" * 42)

    print("\n--- Consumer Demand Spotlight ---")
    print(f"Most Popular Item   : '{most_frequent_item}'")
    print(f"Total Units Ordered : {max_freq} units")
    print("==========================================")


# --- Execution Run ---
# Execute the pipeline against our inline string dataset
analyze_inline_food_orders(inline_csv_data)
```
